# Understanding Code and Markdown Cells

This notebook demonstrates the difference between Code cells and Markdown cells in Jupyter Notebook.


## Purpose of This Notebook

- To demonstrate how Code cells execute Python
- To demonstrate how Markdown cells explain logic
- To structure a notebook clearly and professionally

```python
a = 8
b = 4
result = a * b
result
```

### Explanation of the Above Code

In the previous Code cell:
- We created two variables (a and b)
- Multiplied them
- Displayed the result

This demonstrates that Code cells are used for computation.

```python
message = "Jupyter Notebooks combine code and explanation."
print(message)
```

## When to Use Code vs Markdown

### Use Code cells when:
- Writing Python logic
- Performing calculations
- Running commands

### Use Markdown cells when:
- Explaining what the code does
- Describing results
- Structuring the notebook
- Writing conclusions

Code executes.
Markdown explains.
